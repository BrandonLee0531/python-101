#### 4. Checking Data Types

###### Use `type()`

```python
name = "Alice"
age = 25

print(type(name))  # Output: <class 'str'>
print(type(age))   # Output: <class 'int'>
```

#### 5. Basic Math

```python
a = 11
b = 5


print(a + b)  # Addition: 16
print(a - b)  # Subtraction: 6
print(a * b)  # Multiplication: 55
print(a / b)  # Division: Result is a float: 2.2
print(a // b) # Floor Division: floor division discards the decimal part: 2
print(a % b)  # Modulus: modulo the remainder of the division: 1
print(a ** b) # Exponentiation: 11 raised to the power of 5: 161051
```

#### 6. Comparison Operators

 Operator | Meaning               |
| -------- | --------------------- |
| ==       | Equal                 |
| !=       | Not equal             |
| >        | Greater than          |
| <        | Less than             |
| >=       | Greater than or equal |
| <=       | Less than or equal    |

```python
print(10 == 10)  # True
print(10 != 5)   # True
print(10 > 5)    # True
print(10 < 5)    # False
```

#### 7. Logical Operators

```python
| Operator | Meaning                              |
|----------|--------------------------------------|
| and      | Both conditions must be true         |
| or       | At least one condition must be true  |
| not      | Negates the condition                |
```
